```python
import pandas as pd
import os

# 1. Load the dataset with safety check
file_path = '../data/orders_dataset.csv'

if os.path.exists(file_path):
    df = pd.read_csv(file_path)
    print("✅ System: Dataset loaded successfully.")
else:
    print("❌ Error: Dataset file not found. Please check the /data folder.")

# 2. KPI Calculations using your actual columns: 'total_sales' and 'price'
total_revenue = df['total_sales'].sum()
total_estimated_cost = df['price'].sum()
global_margin = (total_revenue - total_estimated_cost) / total_revenue

print("--- FINAL SALES REPORT ---")
print(f"💰 Total Revenue: ${total_revenue:,.2f} USD")
print(f"📊 Global Gross Margin: {global_margin:.2%}")

# 3. Top Performer Analysis
top_product = df.groupby('product')['total_sales'].sum().idxmax()
print(f"🏆 Top-Selling Product: {top_product}")
```
