```python
import pandas as pd
import os
import matplotlib.pyplot as plt

file_path = '../data/orders_dataset.csv'

if os.path.exists(file_path):
    df = pd.read_csv(file_path, sep='\t')
    df.columns = df.columns.str.strip().str.lower()
    print("✅ System: Dataset loaded successfully.")
else:
    print("❌ Error: Dataset file not found.")

total_revenue = df['revenue'].sum()
total_profit = total_revenue * 0.31

product_sales = df.groupby('product')['revenue'].sum().sort_values(ascending=False)

plt.figure(figsize=(10, 5))
product_sales.plot(kind='bar', color='royalblue')
plt.title('Total Revenue by Product (Python Verification)')
plt.ylabel('Revenue ($)')
plt.xlabel('Product')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()

plt.show()

print(f"\nTotal Revenue: ${total_revenue:,.2f}")
print(f"Adjusted Profit (31%): ${total_profit:,.2f}")
```
